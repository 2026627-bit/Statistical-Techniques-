```python
#Computation of measures of central tendency

mean=df['VALUE'].mean()
print(mean)
```

```python
median=df['VALUE'].median()
print(median)
```

```python
mode=df['VALUE'].mode()
print(mode)
```

```python
#Plot the distribution

data=df['VALUE']

sns.distplot(data, bins=10, hist=True, kde=True, label='VALUE (in Millimetres)')
```

```python
df['VALUE'].min()
```

```python
df['VALUE'].max()
```

```python
#RANGE

df['VALUE'].max()-df['VALUE'].min()
```

```python
#VARIANCE

df['VALUE'].var()
```

```python
#STANDAR DEVIATION

df['VALUE'].std()
```

```python
#MEDIAN (Q2 OR 50TH PERCENTILE)

Q2=df['VALUE'].quantile(0.5)
Q2
```

### Q3 OR 75th Percentile

```python
Q3=df['VALUE'].quantile(0.75)
Q3
```

```python
Q1=df['VALUE'].quantile(0.25)
Q1
```

```python
#INTERQUARTILE RANGE

IQR=Q3-Q1
IQR
```

```python
#DRAW BOXPLOT
sns.boxplot(x=df['VALUE'])
plt.title('VALUES OF PRECIPITATION')
plt.xlabel('Millimetres')
plt.show()
```

```python
#Computation of measures of shape of distribution
#SKEWNESS

df['VALUE'].skew()
```

### KURTOSIS

```python
df['VALUE'].kurt()
```

```python
sns.histplot(df['VALUE'], kde=True)
plt.xlabel('Precipitation in milimeters')
plt.title('Precipitation')
plt.show()
```
